```python
teams = list(df["HomeTeam"].unique())
print("Teams in the dataset:", teams)
```

```python
results = []
for team in teams:
    home_games = df[df["HomeTeam"] == team]
    away_games = df[df["AwayTeam"] == team]

    goals_for = home_games["FTHG"].sum() + away_games["FTAG"].sum()
    xg_for = home_games["HxG"].sum() + away_games["AxG"].sum()
    goals_against = home_games["FTAG"].sum() + away_games["FTHG"].sum()
    xg_against = home_games["AxG"].sum() + away_games["HxG"].sum()

    results.append({
        "Team": team,
        "Matches": len(home_games) + len(away_games),
        "Goals": goals_for,
        "xG": round(xg_for, 2),
        "Goals against": goals_against,
        "xG against": round(xg_against, 2),
        "Attack diff": round(goals_for - xg_for, 2),       # above 0 = scoring more than expected
        "Defence diff": round(xg_against - goals_against, 2),  # above 0 = conceding less than expected
    })

table = pd.DataFrame(results)
table = table.sort_values("Attack diff", ascending=False)
print(table)
```
